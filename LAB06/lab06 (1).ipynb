## 1. Контейнер vs Итерируемый объект vs Итератор vs Генератор

### 1.1 Контейнеры

**Контейнер** - объект, который содержит другие объекты (списки, кортежи, словари, множества).

```python
# Примеры контейнеров
list_container = [1, 2, 3, 4, 5]
tuple_container = (1, 2, 3)
dict_container = {'a': 1, 'b': 2}
set_container = {1, 2, 3, 4}

print(f"Список: {list_container}")
print(f"Кортеж: {tuple_container}")
print(f"Словарь: {dict_container}")
print(f"Множество: {set_container}")
```

### 1.2 Итерируемые объекты

**Итерируемый объект** - объект, который может возвращать элементы по одному.

```python
# Проверка итерируемости
def is_iterable(obj):
    try:
        iter(obj)
        return True
    except TypeError:
        return False

print(f"Список итерируемый: {is_iterable([1, 2, 3])}")
print(f"Число итерируемое: {is_iterable(5)}")
print(f"Строка итерируемая: {is_iterable('hello')}")

# Итерация по итерируемому объекту
for char in "Python":
    print(char, end=' ')
```

### 1.3 Итераторы

**Итератор** - объект, который возвращает следующий элемент при вызове `next()`.

```python
# Создание и использование итератора
my_list = [1, 2, 3]
my_iterator = iter(my_list)

print(f"Первый элемент: {next(my_iterator)}")
print(f"Второй элемент: {next(my_iterator)}")
print(f"Третий элемент: {next(my_iterator)}")

# Попытка получить следующий элемент вызовет StopIteration
try:
    print(f"Четвертый элемент: {next(my_iterator)}")
except StopIteration:
    print("Итератор исчерпан!")
```

### 1.4 Функции enumerate, zip, sorted, reversed

```python
# enumerate - возвращает пары (индекс, элемент)
fruits = ['apple', 'banana', 'cherry']
print("enumerate:")
for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")

# zip - объединяет несколько итерируемых объектов
names = ['Emma', 'John', 'Michael']
ages = [25, 30, 35]
print("\nzip:")
for name, age in zip(names, ages):
    print(f"{name}: {age} лет")

# sorted vs sort
numbers = [3, 1, 4, 1, 5, 9, 2]
sorted_numbers = sorted(numbers)  # возвращает новый список
numbers_copy = numbers.copy()
numbers_copy.sort()  # изменяет исходный список
print(f"\nИсходный список: {numbers}")
print(f"sorted(): {sorted_numbers}")
print(f"После sort(): {numbers_copy}")

# reversed vs reverse
reversed_numbers = list(reversed(numbers))  # возвращает новый список
numbers_copy2 = numbers.copy()
numbers_copy2.reverse()  # изменяет исходный список
print(f"\nreversed(): {reversed_numbers}")
print(f"После reverse(): {numbers_copy2}")
```

### 2.3 Документирование функций с type annotations

```python
# Функция с документацией по стандарту PEP 257 и аннотациями типов
def calculate_area(radius: float) -> float:
    """
    Вычисляет площадь круга по заданному радиусу.
    
    Args:
        radius: Радиус круга. Должен быть положительным числом.
        
    Returns:
        Площадь круга, вычисленная по формуле πr².
        
    Raises:
        ValueError: Если радиус отрицательный.
        
    Examples:
        >>> calculate_area(5)
        78.53981633974483
    """
    if radius < 0:
        raise ValueError("Радиус не может быть отрицательным")
    
    import math
    return math.pi * radius ** 2

# Аннотации типов (PEP 484)
def process_data(data: list[str], reverse: bool = False) -> tuple[int, list[str]]:
    """
    Обрабатывает список строк.
    
    Args:
        data: Список строк для обработки
        reverse: Если True, возвращает список в обратном порядке
        
    Returns:
        Кортеж из количества элементов и обработанного списка
    """
    count = len(data)
    if reverse:
        data = data[::-1]
    return count, data

# Использование функций
area = calculate_area(5)
print(f"Площадь круга с радиусом 5: {area:.2f}")

count, processed = process_data(["apple", "banana", "cherry"], reverse=True)
print(f"Количество: {count}, Обработанный список: {processed}")

# Просмотр аннотаций
print(f"\nАннотации calculate_area: {calculate_area.__annotations__}")
```

**Задача 2.2:** Напишите функцию с документацией, которая проверяет палиндром

```python
def is_palindrome(text):
    """
    Проверяет, является ли переданный текст палиндромом.
    
    Args:
        text (str): Текст для проверки на палиндром
        
    Returns:
        bool: True если текст является палиндромом, False в противном случае
        
    Examples:
        >>> is_palindrome("А роза упала на лапу Азора")
        True
        >>> is_palindrome("hello")
        False
    """
    cleaned_text = ''.join(char.lower() for char in text if char.isalnum())
    return cleaned_text == cleaned_text[::-1]


print(is_palindrome("arvvvra"))
```

**Задача 2.3:** Создайте функцию с аннотациями типов для вычисления среднего значения

```python
def calculate_average(numbers: list[int]):
    """
    Вычисляет среднее арифметическое значение списка чисел.
    
    Args:
        numbers: Список чисел (целых или вещественных)
        
    Returns:
        Среднее арифметическое значение
        
    Raises:
        ValueError: Если передан пустой список
        
    Examples:
        >>> calculate_average([1.5, 2.5, 3.5])
        2.5
    """
    if not numbers:
        raise ValueError("Список чисел не может быть пустым")
    
    return sum(numbers) / len(numbers)

# Примеры использования
print(calculate_average([1, 2, 3, 4, 5]))      # 3.0
print(calculate_average([10, 20, 30]))         # 20.0
```

**Задача 2.4:** Напишите функцию, которая возвращает кортеж из минимума и максимума

```python
def find_min_max(numbers: list) -> tuple:
    """
    Находит минимальное и максимальное значение в списке чисел.
    
    Args:
        numbers: Список чисел для анализа
        
    Returns:
        Кортеж в формате (минимум, максимум)
        
    Raises:
        ValueError: Если передан пустой список
        
    Examples:
        >>> find_min_max([3, 1, 4, 1, 5, 9, 2])
        (1, 9)
        >>> find_min_max([-5, 0, 15, 3])
        (-5, 15)
    """
    if not numbers:
        raise ValueError("Список чисел не может быть пустым")
    
    min_value = min(numbers)
    max_value = max(numbers)
    
    return (min_value, max_value)
    
result = find_min_max([-10, 0, 25, 100, -5])
print(f"Минимум и максимум: {result}")
```

**Задача 2.5:** Создайте функцию поиска элемента в списке с полной документацией по PEP 257

```python
from typing import Any, List, Optional
def find_element_index(collection: List[Any], target: Any, case_sensitive: bool = True) -> Optional[int]:
    """
    Поиск элемента в списке.
    
    Функция выполняет поиск в коллекции
    и возвращает индекс его первого вхождения. Если элемент не найден,
    возвращает None.
    
    Args:
        collection: Коллекция элементов для поиска
        target: Элемент, который необходимо найти
        case_sensitive: Флаг чувствительности к регистру (для строк).
                       По умолчанию True.
        
    Returns:
        Индекс первого вхождения элемента или None, если элемент не найден.
        
    Raises:
        TypeError: Если collection не является итерируемым объектом.
        
    Examples:
        >>> find_element_index([1, 2, 3, 4, 5], 3)
        2
        >>> find_element_index(['apple', 'banana', 'orange'], 'banana')
        1
        >>> find_element_index(['Hello', 'World'], 'hello')
        None
        >>> find_element_index(['Hello', 'World'], 'hello', case_sensitive=False)
        0
    """
    try:
        if not case_sensitive and all(isinstance(item, str) for item in collection) and isinstance(target, str):
            target_lower = target.lower()
            for index, item in enumerate(collection):
                if isinstance(item, str) and item.lower() == target_lower:
                    return index
        else:
            for index, item in enumerate(collection):
                if item == target:
                    return index
        
        return None
        
    except TypeError as e:
        raise TypeError("Параметр collection должен быть итерируемым") from e

numbers = [10, 20, 30, 40, 50]
index = find_element_index(numbers, 30)
print(f"Индекс числа 30: {index}")  # 2

fruits = ['apple', 'banana', 'orange', 'grape']
index = find_element_index(fruits, 'orange')
print(f"Индекс 'orange': {index}")  # 2

index = find_element_index(numbers, 100)
print(f"Индекс числа 100: {index}")
```

Примеры вызова:
- `format_text("текст")` → "текст              "
- `format_text("текст", width=10)` → "текст      "
- `format_text("текст", align='center', fill_char='-')` → "-------текст-------"

```python
def format_text(text, alignment='left', width=20, fill_char=' '):
    if alignment == 'left':
        return text.ljust(width, fill_char)
    elif alignment == 'right':
        return text.rjust(width, fill_char)
    elif alignment == 'center':
        return text.center(width, fill_char)
    else:
        raise ValueError("Выравнивание должно быть 'left', 'right' или 'center'")
print(f"{format_text("текст")}")                
print(f"{format_text("текст", width = 10)}")       
print(f"{format_text('текст', alignment='center',fill_char='-')}") 
```

**Задача 3.5:** Создайте функцию `filter_data`, которая принимает список чисел и булевые аргументы по умолчанию: filter_even=True, filter_positive=True. Функция должна фильтровать числа в соответствии с параметрами.

```python
def filter_data(numbers: List[int],filter_even:bool = True,filter_positive:bool = True):
    filtered_numbers = []
    
    for number in numbers:
        even_condition = not filter_even or (filter_even and number % 2 != 0)
        positive_condition = not filter_positive or (filter_positive and number > 0)
        if even_condition and positive_condition:
            filtered_numbers.append(number)
    
    return filtered_numbers
test_data = [-3, -2, -1, 0, 1, 2, 3, 4, 5]
print("Фильтрация данных:")
print(f"Исходный список: {test_data}")
print(f"Только нечетные положительные: {filter_data(test_data)}")  # [1, 3, 5]
print(f"Все положительные: {filter_data(test_data, filter_even=False)}")  # [1, 2, 3, 4, 5]
print(f"Только нечетные: {filter_data(test_data, filter_positive=False)}")  # [-3, -1, 1, 3, 5]
print(f"Без фильтрации: {filter_data(test_data, filter_even=False, filter_positive=False)}")  # Все числа
```

### 4.2 Ключевые аргументы

```python
# Функция с ключевыми аргументами
def send_email(to, subject, body, cc=None, bcc=None, reply_to=None, priority="normal"):
    """
    Отправляет электронное письмо.
    
    Args:
        to: Адрес получателя
        subject: Тема письма
        body: Текст письма
        cc: Копия
        bcc: Скрытая копия
        reply_to: Адрес для ответа
        priority: Приоритет письма
    """
    email_info = f"To: {to}\nSubject: {subject}\nBody: {body}\nPriority: {priority}"
    
    if cc:
        email_info += f"\nCC: {cc}"
    if bcc:
        email_info += f"\nBCC: {bcc}"
    if reply_to:
        email_info += f"\nReply-To: {reply_to}"
    
    return email_info

# Использование ключевых аргументов
email2 = send_email(
    to="user@example.com",
    subject="Важное сообщение",
    body="Пожалуйста, ответьте ASAP",
    cc="manager@example.com",
    reply_to="support@example.com",
    priority="high"
)
print("Письмо с дополнительными параметрами:")
print(email2)
```

**Задача 4.5:** Создайте функцию `calculate_discount`, где порядок именованных аргументов не важен. Параметры: price, discount_percent, tax_rate.

```python
def calculate_discount(*, price, discount_percent=0, tax_rate=0):
    if price < 0:
        raise ValueError("Цена не может быть отрицательной")
    if discount_percent < 0 or discount_percent > 100:
        raise ValueError("Скидка должна быть в диапазоне от 0 до 100%")
    if tax_rate < 0:
        raise ValueError("Налог не может быть отрицательным")    
    discount_amount = price * (discount_percent / 100)
    price_after_discount = price - discount_amount    
    tax_amount = price_after_discount * (tax_rate / 100)
    final_price = price_after_discount + tax_amount
    
    return {
        'original_price': price,
        'discount_percent': discount_percent,
        'discount_amount': round(discount_amount, 2),
        'price_after_discount': round(price_after_discount, 2),
        'tax_rate': tax_rate,
        'tax_amount': round(tax_amount, 2),
        'final_price': round(final_price, 2)
    }
print(calculate_discount(price=1000, discount_percent=10, tax_rate=20))
print(calculate_discount(discount_percent=15, price=500, tax_rate=10))
print(calculate_discount(tax_rate=5, price=200, discount_percent=5))
print(calculate_discount(price=1500))  # Только обязательный параметр
```

```python
def process_data(*args, **kwargs):
    multiplier = kwargs.get('multiplier', 1)
    round_result = kwargs.get('round_result', False)
    decimals = kwargs.get('decimals', 2)
    filter_positive = kwargs.get('filter_positive', False)
    
    result = []
    
    for number in args:
        processed = number * multiplier
    
        if filter_positive and processed <= 0:
            continue
            
        if round_result:
            processed = round(processed, decimals)
            
        result.append(processed)
    
    return result

numbers = [1.234, 2.567, 3.891, -4.123]

result1 = process_data(*numbers)
print("Без параметров:", result1)

result2 = process_data(*numbers, multiplier=2.5, round_result=True)
print("С множителем 2.5 и округлением:", result2)

result3 = process_data(*numbers, filter_positive=True, round_result=True, decimals=1)
print("Только положительные, округленные:", result3)
```

## 8. Пользовательские генераторы

### 8.1 Генераторы с yield

```python
# Простой генератор
def countdown(n):
    """Генератор обратного отсчета."""
    print("Начало отсчета")
    while n > 0:
        yield n
        n -= 1
    print("Отсчет завершен!")

# Использование генератора
print("Обратный отсчет:")
for number in countdown(5):
    print(number)

# Генератор можно использовать вручную
print("\nРучное использование:")
counter = countdown(3)
print(next(counter))
print(next(counter))
print(next(counter))

# После завершения генератор вызывает StopIteration
try:
    print(next(counter))
except StopIteration:
    print("Генератор завершен")
```

### 8.2 Преимущества генераторов

```python
# Сравнение подхода с списком и генератором
def squares_list(n):
    """Возвращает список квадратов."""
    result = []
    for i in range(n):
        result.append(i ** 2)
    return result

def squares_generator(n):
    """Генератор квадратов."""
    for i in range(n):
        yield i ** 2

# Для больших n генератор экономит память
n = 1000000

print("Создание списка квадратов...")
# squares = squares_list(n)  # Занимает много памяти
print("Список создан (закомментировано для экономии памяти)")

print("Создание генератора квадратов...")
squares_gen = squares_generator(n)
print("Генератор создан")

# Получение первых 5 значений из генератора
print("Первые 5 квадратов:")
for i, square in enumerate(squares_gen):
    if i >= 5:
        break
    print(square)

print("Генератор продолжает работать с того места, где остановился...")
```

```python
# Простой класс
class Dog:
    """Простой класс, представляющий собаку."""
    
    # Атрибут класса (общий для всех экземпляров)
    species = "Canis familiaris"
    
    def __init__(self, name, age):
        """Конструктор класса.
        
        Args:
            name (str): Имя собаки
            age (int): Возраст собаки
        """
        # Атрибуты экземпляра (уникальные для каждого объекта)
        self.name = name
        self.age = age
    
    # Метод экземпляра
    def bark(self):
        """Собака лает."""
        return f"{self.name} говорит: Гав!"
    
    def get_info(self):
        """Возвращает информацию о собаке."""
        return f"{self.name}, {self.age} лет, вид: {self.species}"

# Создание объектов (экземпляров класса)
dog1 = Dog("Buddy", 3)
dog2 = Dog("Max", 5)

# Доступ к атрибутам
print(f"dog1.name: {dog1.name}")
print(f"dog2.name: {dog2.name}")
print(f"dog1.species: {dog1.species}")
print(f"dog2.species: {dog2.species}")

# Вызов методов
print(dog1.bark())
print(dog2.bark())
print(dog1.get_info())
print(dog2.get_info())

# Изменение атрибутов
dog1.age = 4
print(f"После изменения: {dog1.get_info()}")
```

### 9.2 Специальные методы

```python
class Book:
    """Класс, представляющий книгу."""
    
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
    
    # Строковое представление
    def __str__(self):
        return f"'{self.title}' by {self.author}"
    
    # Формальное строковое представление
    def __repr__(self):
        return f"Book('{self.title}', '{self.author}', {self.pages})"
    
    # Длина книги
    def __len__(self):
        return self.pages

# Использование
book1 = Book("War and Peace", "Leo Tolstoy", 1225)
book2 = Book("Crime and Punishment", "Fyodor Dostoevsky", 671)

print(f"str(book1): {str(book1)}")
print(f"repr(book1): {repr(book1)}")
print(f"Длина книги: {len(book1)} страниц")
```

**Задача 9.2:** Реализуйте класс «Банковский счет» с методами пополнения и снятия средств

```python
class Bank_account:
    def __init__(self,account_holder, initial_balance=0):
        self.account_holder = account_holder
        self._balance = initial_balance
        self._transaction_history = []
    
    
    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Сумма пополнения должна быть положительной")
        
        self._balance += amount
        self._transaction_history.append(f"Пополнение: +{amount}")
        return f"Счет пополнен на {amount}. Текущий баланс: {self._balance}"
    
    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Сумма снятия должна быть положительной")
        
        if amount > self._balance:
            raise ValueError("Недостаточно средств на счете")
        
        self._balance -= amount
        self._transaction_history.append(f"Снятие: -{amount}")
        return f"Со счета снято {amount}. Текущий баланс: {self._balance}"
    
    def get_balance(self):
        """Получить текущий баланс"""
        return self._balance
    
    def get_transaction_history(self):
        """Получить историю операций"""
        return self._transaction_history.copy()

account = Bank_account("Иван Иванов", 1010)
print(f"Владелец счета: {account.account_holder}")
print(f"Начальный баланс: {account.get_balance()}")

print(account.deposit(500))
print(account.withdraw(200))
print(account.deposit(1000))

print("\nИстория операций:")
for transaction in account.get_transaction_history():
    print(f" - {transaction}")

print(f"\nИтоговый баланс: {account.get_balance()}")
```

**Задача 9.3:** Создайте класс с использованием свойств для валидации данных

```python
class Person:
    def __init__(self, name, age, email):
        self._name = name
        self._age = age
        self._email = email
    
    @property
    def name(self):
        return self._name
    
    @name.setter
    def name(self, value):
        if not value or not isinstance(value, str):
            raise ValueError("Имя должно быть непустой строкой")
        if len(value) < 2:
            raise ValueError("Имя должно содержать минимум 2 символа")
        self._name = value
    
    @property
    def age(self):
        return self._age
    
    @age.setter
    def age(self, value):
        if not isinstance(value, int):
            raise ValueError("Возраст должен быть целым числом")
        if value < 0 or value > 100:
            raise ValueError("Возраст должен быть в диапазоне от 0 до 100")
        self._age = value
    
    @property
    def email(self):
        return self._email
    
    @email.setter
    def email(self, value):
        if not isinstance(value, str) or "@" not in value:
            raise ValueError("Email должен быть строкой и содержать символ @")
        self._email = value
    
    def __str__(self):
        return f"Person(name='{self.name}', age={self.age}, email='{self.email}')"

try:
    person = Person("Анна", 25, "anna@gmail.com")
    print(person)
    
    person.name = "Анна Петрова"
    person.age = 26
    person.email = "anna.petrovagmail.com"
    print("После изменений:", person)
        
except ValueError as e:
    print(f"Ошибка валидации: {e}")
```

**Задача 9.5:** Создайте класс `MathUtils` со статическими методами `add`, `multiply` и методом класса `create_from_string`, который создает объект из строки с числами.

```python
class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b
    
    @staticmethod
    def multiply(a, b):
        return a * b
    
    @classmethod
    def create_from_string(cls, string_data):
        try:
            numbers = [float(x.strip()) for x in string_data.split(',')]
            return cls(*numbers)
        except ValueError as e:
            raise ValueError(f"Некорректный формат строки: {string_data}") from e
    
    def __init__(self, *numbers):
        self.numbers = numbers
    
    def sum_all(self):
        return sum(self.numbers)
    
    def product_all(self):
        result = 1
        for num in self.numbers:
            result *= num
        return result
    
    def __str__(self):
        return f"MathUtils({', '.join(map(str, self.numbers))})"

print("=== Статические методы ===")
print(f"Сложение: 5 + 3 = {MathUtils.add(5, 3)}")
print(f"Умножение: 4 * 7 = {MathUtils.multiply(4, 7)}")

result = MathUtils.add(10.5, 20.3)
print(f"Сложение дробных: 10.5 + 20.3 = {result}")

print("\n=== Метод класса create_from_string ===")
try:
    math_obj1 = MathUtils.create_from_string("5, 3, 2")
    print(f"Создан объект: {math_obj1}")
    print(f"Сумма чисел: {math_obj1.sum_all()}")
    print(f"Произведение чисел: {math_obj1.product_all()}")
    
    math_obj2 = MathUtils.create_from_string("1.5, 2.5")
    print(f"\nСоздан объект: {math_obj2}")
    print(f"Сумма: {math_obj2.sum_all()}")
    

    
except ValueError as e:
    print(f"Ошибка: {e}")
```

### 10.1 Инкапсуляция

```python
class BankAccount:
    """Класс банковского счета с инкапсуляцией."""
    
    def __init__(self, account_holder, initial_balance=0):
        self.account_holder = account_holder
        self._balance = initial_balance  # Защищенный атрибут
        self.__account_id = hash(account_holder)  # Приватный атрибут
    
    # Публичные методы для взаимодействия
    def deposit(self, amount):
        """Пополнение счета."""
        if amount > 0:
            self._balance += amount
            return True
        return False
    
    def withdraw(self, amount):
        """Снятие со счета."""
        if 0 < amount <= self._balance:
            self._balance -= amount
            return True
        return False
    
    def get_balance(self):
        """Получение баланса."""
        return self._balance

# Использование
account = BankAccount("Alex Johnson", 1000)

print(f"Владелец: {account.account_holder}")
print(f"Баланс: {account.get_balance()}")

account.deposit(500)
print(f"После пополнения: {account.get_balance()}")

account.withdraw(200)
print(f"После снятия: {account.get_balance()}")

# Попытка доступа к защищенным и приватным атрибутам
print(f"\nЗащищенный атрибут (_balance): {account._balance}")  # Доступно, но не рекомендуется
# print(f"Приватный атрибут (__account_id): {account.__account_id}")  # Ошибка!
```

### 10.2 Наследование

```python
# Базовый класс
class Animal:
    """Базовый класс для всех животных."""
    
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def speak(self):
        """Животное издает звук."""
        raise NotImplementedError("Подклассы должны реализовать этот метод")
    
    def get_info(self):
        """Возвращает информацию о животном."""
        return f"{self.name}, {self.age} лет"

# Производные классы
class Dog(Animal):
    """Класс собаки."""
    
    def __init__(self, name, age, breed):
        super().__init__(name, age)  # Вызов конструктора родительского класса
        self.breed = breed
    
    def speak(self):
        """Собака лает."""
        return "Гав!"
    
    def get_info(self):
        """Возвращает информацию о собаке."""
        return f"{super().get_info()}, порода: {self.breed}"

class Cat(Animal):
    """Класс кошки."""
    
    def speak(self):
        """Кошка мяукает."""
        return "Мяу!"

# Использование
animals = [
    Dog("Buddy", 3, "Labrador"),
    Cat("Whiskers", 2)
]

for animal in animals:
    print(f"{animal.get_info()} говорит: {animal.speak()}")

# Проверка типов
print(f"\nПроверка типов:")
print(f"Buddy - Animal: {isinstance(animals[0], Animal)}")
print(f"Buddy - Dog: {isinstance(animals[0], Dog)}")
print(f"Buddy - Cat: {isinstance(animals[0], Cat)}")
```

### 10.3 Абстрактные классы

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    """Абстрактный класс геометрической фигуры."""
    
    def __init__(self, color):
        self.color = color
    
    @abstractmethod
    def area(self):
        """Вычисляет площадь фигуры."""
        pass
    
    @abstractmethod
    def perimeter(self):
        """Вычисляет периметр фигуры."""
        pass
    
    def get_info(self):
        """Возвращает информацию о фигуре."""
        return f"Фигура цвета {self.color}, площадь: {self.area():.2f}, периметр: {self.perimeter():.2f}"

class Rectangle(Shape):
    """Класс прямоугольника."""
    
    def __init__(self, color, width, height):
        super().__init__(color)
        self.width = width
        self.height = height
    
    def area(self):
        """Площадь прямоугольника."""
        return self.width * self.height
    
    def perimeter(self):
        """Периметр прямоугольника."""
        return 2 * (self.width + self.height)

class Circle(Shape):
    """Класс круга."""
    
    def __init__(self, color, radius):
        super().__init__(color)
        self.radius = radius
    
    def area(self):
        """Площадь круга."""
        import math
        return math.pi * self.radius ** 2
    
    def perimeter(self):
        """Периметр круга (длина окружности)."""
        import math
        return 2 * math.pi * self.radius

# Использование
shapes = [
    Rectangle("красный", 5, 3),
    Circle("синий", 4)
]

for shape in shapes:
    print(shape.get_info())

# Попытка создать экземпляр абстрактного класса
try:
    shape = Shape("черный")
except TypeError as e:
    print(f"\nОшибка: {e}")
```

```python
class Employee:
    
    def __init__(self, name, employee_id, salary):
        self.name = name
        self.employee_id = employee_id
        self.salary = salary
    
    def calculate_bonus(self):
        return self.salary * 0.1
    
    def get_info(self):
        return f"{self.name} (ID: {self.employee_id})"


class Manager(Employee):
    
    def __init__(self, name, employee_id, salary, team_size):
        super().__init__(name, employee_id, salary)
        self.team_size = team_size
    
    def calculate_bonus(self):
        return self.salary * 0.15 + self.team_size * 1000
    
    def conduct_meeting(self):
        return f"{self.name} проводит совещание с командой из {self.team_size} человек"


class Developer(Employee):
    
    def __init__(self, name, employee_id, salary, programming_language):
        super().__init__(name, employee_id, salary)
        self.programming_language = programming_language
    
    def calculate_bonus(self):
        return self.salary * 0.12
    
    def write_code(self):
        return f"{self.name} пишет код на {self.programming_language}"


class SalesPerson(Employee):
    
    def __init__(self, name, employee_id, salary, sales_volume=0):
        super().__init__(name, employee_id, salary)
        self.sales_volume = sales_volume
    
    def calculate_bonus(self):
        return self.sales_volume * 0.05
    
    def make_sale(self, amount):
        self.sales_volume += amount
        return f"{self.name} совершил продажу на {amount}"

# Пример использования
print("ИЕРАРХИЯ СОТРУДНИКОВ")
employees = [
    Manager("Иван Иванов", "M001", 100000, 5),
    Developer("Петр Петров", "D001", 80000, "Python"),
    SalesPerson("Мария Сидорова", "S001", 60000, 500000)
]

for employee in employees:
    print(f"{employee.get_info()}")
    print(f"Бонус: {employee.calculate_bonus():.0f} руб.")
    if isinstance(employee, Manager):
        print(f"Действие: {employee.conduct_meeting()}")
    elif isinstance(employee, Developer):
        print(f"Действие: {employee.write_code()}")
    elif isinstance(employee, SalesPerson):
        print(f"Действие: {employee.make_sale(10000)}")
    print()
```

**Задача 10.2:** Реализуйте абстрактный класс для игровых персонажей

```python
from abc import ABC, abstractmethod

class GameCharacter(ABC):
    
    def __init__(self, name, health, level):
        self.name = name
        self.health = health
        self.level = level
    
    @abstractmethod
    def attack(self):
        pass
    
    @abstractmethod
    def special_ability(self):
        pass
    
    def take_damage(self, damage):
        self.health -= damage
        if self.health < 0:
            self.health = 0
        return f"{self.name} получает {damage} урона. Здоровье: {self.health}"
    
    def is_alive(self):
        return self.health > 0
    
    def __str__(self):
        return f"{self.name} (Уровень: {self.level}, Здоровье: {self.health})"


class Warrior(GameCharacter):
    
    def __init__(self, name, health, level, strength):
        super().__init__(name, health, level)
        self.strength = strength
    
    def attack(self):
        return f"{self.name} наносит мечом {self.strength * 2} урона"
    
    def special_ability(self):
        return f"{self.name} использует 'Берсерк'! Сила увеличена в 2 раза"
    
    def block(self):
        return f"{self.name} блокирует атаку щитом"


class Mage(GameCharacter):
    
    def __init__(self, name, health, level, mana):
        super().__init__(name, health, level)
        self.mana = mana
    
    def attack(self):
        return f"{self.name} бросает огненный шар на {self.level * 3} урона"
    
    def special_ability(self):
        if self.mana >= 50:
            self.mana -= 50
            return f"{self.name} использует 'Ледяную бурю'! Мана: {self.mana}"
        else:
            return f"Недостаточно маны для использования способности"
    
    def meditate(self):
        self.mana += 30
        return f"{self.name} медитирует. Мана восстановлена до {self.mana}"


class Archer(GameCharacter):
    
    def __init__(self, name, health, level, accuracy):
        super().__init__(name, health, level)
        self.accuracy = accuracy
    
    def attack(self):
        damage = self.level * 2 + self.accuracy
        return f"{self.name} выпускает стрелу на {damage} урона"
    
    def special_ability(self):
        return f"{self.name} использует 'Прицельный выстрел'! Точность +50%"
    
    def dodge(self):
        return f"{self.name} уворачивается от атаки"

print("ИГРОВЫЕ ПЕРСОНАЖИ")
characters = [
    Warrior("Боромир", 100, 5, 15),
    Mage("Гэндальф", 80, 8, 100),
    Archer("Леголас", 90, 6, 20)
]

for character in characters:
    print(character)
    print(f"Атака: {character.attack()}")
    print(f"Способность: {character.special_ability()}")
    
    if isinstance(character, Warrior):
        print(f"Уникальное: {character.block()}")
    elif isinstance(character, Mage):
        print(f"Уникальное: {character.meditate()}")
    elif isinstance(character, Archer):
        print(f"Уникальное: {character.dodge()}")
    
    print(f"Жив: {'Да' if character.is_alive() else 'Нет'}")
    print()
```

**Задача 10.3:** Создайте класс `SmartDevice`, который наследуется от классов `Phone` и `Tablet`. Класс `Phone` имеет метод `call()`, класс `Tablet` - метод `browse()`. Класс `SmartDevice` должен иметь оба метода.

```python
class Phone:
    
    def __init__(self, phone_number):
        self.phone_number = phone_number
    
    def call(self, number):
        return f"Звоню на номер {number} с {self.phone_number}"
    
    def send_sms(self, number, message):
        return f"Отправляю SMS на {number}: '{message}'"


class Tablet:
    
    def __init__(self, screen_size):
        self.screen_size = screen_size
    
    def browse(self, url):
        return f"Открываю в браузере: {url}"
    
    def watch_video(self, video_title):
        return f"Смотрю видео '{video_title}' на экране {self.screen_size}\""


class SmartDevice(Phone, Tablet):
    
    def __init__(self, phone_number, screen_size, model):
        Phone.__init__(self, phone_number)
        Tablet.__init__(self, screen_size)
        self.model = model
        self.battery_level = 100
    
    def use_app(self, app_name):
        self.battery_level -= 5
        return f"Запускаю приложение {app_name}. Батарея: {self.battery_level}%"
    
    def charge(self):
        self.battery_level = 100
        return f"Устройство заряжено до {self.battery_level}%"
    
    def get_info(self):
        return f"{self.model} (Номер: {self.phone_number}, Экран: {self.screen_size}\")"

print("УМНОЕ УСТРОЙСТВО")
smartphone = SmartDevice("+7-123-456-78-90", 6.7, "SuperPhone X")

print(smartphone.get_info())
print(f"Телефонные функции: {smartphone.call('+7-987-654-32-10')}")
print(f"Планшетные функции: {smartphone.browse('https://example.com')}")
print(f"Гибридные функции: {smartphone.use_app('Instagram')}")
print(f"SMS: {smartphone.send_sms('+7-111-222-33-44', 'Привет!')}")
print(f"Видео: {smartphone.watch_video('Новый трейлер')}")
print(f"Зарядка: {smartphone.charge()}")

print(f"\nПроверка наследования:")
print(f"SmartDevice является Phone: {isinstance(smartphone, Phone)}")
print(f"SmartDevice является Tablet: {isinstance(smartphone, Tablet)}")
print(f"Порядок разрешения методов (MRO): {SmartDevice.__mro__}")
```

```python
class Calculator:
    
    def __init__(self):
        self.last_result = 0
    
    def add(self, a, b):
        self.last_result = a + b
        return self.last_result
    
    def subtract(self, a, b):
        self.last_result = a - b
        return self.last_result
    
    def multiply(self, a, b):
        self.last_result = a * b
        return self.last_result
    
    def divide(self, a, b):
        if b == 0:
            raise ValueError("Деление на ноль невозможно")
        self.last_result = a / b
        return self.last_result
    
    def __str__(self):
        return f"Calculator (последний результат: {self.last_result})"


class AdvancedCalculator(Calculator):
    
    def __init__(self):
        super().__init__()
        self.operation_count = 0
    
    def power(self, base, exponent):
        self.last_result = base ** exponent
        self.operation_count += 1
        return self.last_result
    
    def root(self, number, n=2):
        if number < 0 and n % 2 == 0:
            raise ValueError("Корень четной степени из отрицательного числа не существует")
        self.last_result = number ** (1/n)
        self.operation_count += 1
        return self.last_result
    def add(self, a, b):
        result = super().add(a, b)
        self.operation_count += 1
        return result
    
    def subtract(self, a, b):
        result = super().subtract(a, b)
        self.operation_count += 1
        return result
    
    def multiply(self, a, b):
        result = super().multiply(a, b)
        self.operation_count += 1
        return result
    
    def divide(self, a, b):
        result = super().divide(a, b)
        self.operation_count += 1
        return result
    
    def __str__(self):
        return f"AdvancedCalculator (операций: {self.operation_count}, последний результат: {self.last_result})"


calc = AdvancedCalculator()

print("Базовые операции:")
print(f"5 + 3 = {calc.add(5, 3)}")
print(f"10 - 4 = {calc.subtract(10, 4)}")
print(f"6 * 7 = {calc.multiply(6, 7)}")
print(f"15 / 3 = {calc.divide(15, 3)}")

print("\nПродвинутые операции:")
print(f"2^8 = {calc.power(2, 8)}")
print(f"√16 = {calc.root(16):.2f}")
print(f"∛27 = {calc.root(27, 3):.2f}")

print(f"\n{calc}")
```

**Задача 10.5:** Создайте класс `SecureData` с инкапсуляцией данных. Используйте защищенные и приватные атрибуты для хранения конфиденциальной информации.

```python
class SecureData:
    
    def __init__(self, username, password, email, credit_card=None):
        self.username = username
        self._email = email
        
        self.__password = self._encrypt_password(password)
        self.__credit_card = self._mask_credit_card(credit_card) if credit_card else None
        self.__login_attempts = 0
        self.__is_locked = False
    
    def _encrypt_password(self, password):
        # В реальном приложении здесь должно быть настоящее шифрование
        return f"encrypted_{password}_hash"
    
    def _mask_credit_card(self, card_number):
        if len(card_number) != 16 or not card_number.isdigit():
            raise ValueError("Номер карты должен содержать 16 цифр")
        return f"****-****-****-{card_number[-4:]}"
    
    def verify_password(self, input_password):
        if self.__is_locked:
            raise Exception("Аккаунт заблокирован из-за слишком большого количества попыток")
        
        encrypted_input = self._encrypt_password(input_password)
        if encrypted_input == self.__password:
            self.__login_attempts = 0 
            return True
        else:
            self.__login_attempts += 1
            if self.__login_attempts >= 3:
                self.__is_locked = True
                raise Exception("Аккаунт заблокирован! Слишком много неудачных попыток")
            return False
    
    def get_email(self):
        if self._email:
            parts = self._email.split('@')
            if len(parts) == 2:
                return f"{parts[0][:2]}***@{parts[1]}"
        return self._email
    
    def get_credit_card_info(self):
        return self.__credit_card

    def set_email(self, new_email):
        if '@' not in new_email or '.' not in new_email.split('@')[1]:
            raise ValueError("Некорректный формат email")
        self._email = new_email
    
    def change_password(self, old_password, new_password):
        if not self.verify_password(old_password):
            raise ValueError("Неверный старый пароль")
        
        if len(new_password) < 6:
            raise ValueError("Пароль должен содержать минимум 6 символов")
        
        self.__password = self._encrypt_password(new_password)
        return "Пароль успешно изменен"
    
    def __str__(self):
        return f"SecureData(username='{self.username}', email='{self.get_email()}', attempts={self.__login_attempts})"
    
    def get_security_status(self):
        status = "активен" if not self.__is_locked else "заблокирован"
        return f"Статус: {status}, Попыток входа: {self.__login_attempts}"


print("\nSECURE DATA")
try:
    user_data = SecureData(
        username="ivan_petrov",
        password="secret123",
        email="ivan@example.com",
        credit_card="1234567812345678"
    )
    
    print(f"Создан пользователь: {user_data}")
    print(f"Email (маскированный): {user_data.get_email()}")
    print(f"Карта: {user_data.get_credit_card_info()}")
    print(f"Статус: {user_data.get_security_status()}")
    
    print("\nПроверка пароля:")
    print(f"Правильный пароль: {user_data.verify_password('secret123')}")
    print(f"Неправильный пароль: {user_data.verify_password('wrong')}")
    
    print("\nСмена пароля:")
    user_data.change_password("secret123", "newsecret456")
    print("Пароль изменен успешно!")
    print(f"Новый пароль работает: {user_data.verify_password('newsecret456')}")
    
    print("\nОбновление email:")
    user_data.set_email("ivan.new@company.com")
    print(f"Новый email: {user_data.get_email()}")
    
except Exception as e:
    print(f"Ошибка: {e}")

print("\nЗАЩИТА ОТ ПЕРЕБОРА")
test_user = SecureData("test", "password123", "test@test.com")
try:
    for i in range(5):
        test_user.verify_password("wrong")
        print(f"Попытка {i+1}: неудача")
except Exception as e:
    print(f"Результат: {e}")
```
